### Item a
Dado $p\in\mathcal P_d$ um polinômio qualquer, defina $b_p = (p(x_1),p(x_2),\dots,p(x_m))\in\mathbb R^m$ como o vetor obtido por $L_{X}(p)$. Além disso, seja $q = R_{X,d}(b)\in\mathcal P_d$ o polinômio que minimiza $||L_X(q) - b_p||^2_2$, mas como $b_p = L_X(p)$, tomando $q = p$, segue que

$$||L_X(q) - b_p||^2_2 = ||L_X(q) - L_X(p)||^2_2 = ||L_X(p) - L_X(p)||^2_2 = 0$$

Portanto, $p$ é um minimizador, basta mostrar que é único. Para isso, provemos que $N(L_X) = \set 0$ em $\mathcal P_d$.

Suponha por absurdo que exista $p\neq 0 \in\mathcal P_d$ tal que $L_X(p) = 0$, assim $(p(x_1),\dots,p(x_m)) = (0,\dots,0)$, ou seja, $p(x_i) = 0, \forall i \in$ {$1,\dots, m$}, o que implica que $p$ possui $m$ raízes, mas isso é um absurdo, pois $p$ é um polinômio de grau até $d < d+1\leq m$. Logo, como o núcleo de $L_X : \mathcal P_d \rightarrow \mathbb R^m$ é trivial, $L_X$ é injetiva e, portanto, o minimizador é único.

Com isso, temos que

$$||L_X(q) - L_X(p)||^2_2 = 0 \Leftrightarrow L_X(q) - L_X(p) = 0 \Leftrightarrow L_X(q) = L_X(p) \Leftrightarrow q = p$$

Assim, por fim, temos que $R_{X,d}(L_X(p)) = p$ e como isso vale para todo $p \in \mathcal P_d$, segue que $R_{X,d}\circ L_X = \mathcal Id_{\mathcal P_d}.$

### Item b

Basta provar que $R_{X,d}$ é linear na soma e na multiplicação por escalar.

De fato, temos que $R_{X,d}(b + v) = R_{X,d}(b) + R_{X,d}(v)$, pois dados $b,v \in \mathbb R^m$ defina $p = R_{X,d}(b)$ e $q = R_{X,d}(v)$ polinômios de $\mathcal P_d$, defina ainda $h = p + q$. Assim, uma vez que $L_X$ é uma transformação linear*, $L_X(h) = L_X(p) + L_X(q)$, logo $R_{X,d}(L_X(h)) = R_{X,d}(L_X(p) + L_X(q))$, mas como, pelo item a, $R_{X,d}(L_X(h)) = h = p + q = R_{X,d}(L_X(p)) + R_{X,d}(L_X(q))$, segue que é linear para soma.

Ademais, dado $\alpha \in \mathbb R$, temos que

$$R_{X,d}(\alpha b) = R_{X,d}(L_X(\alpha p) = \alpha p = \alpha R_{X,d}(L_X(p)) = \alpha R_{X,d}(b))$$

Logo também é linear para a multiplicação por escalar.

*Dados $f,g \in \mathcal F$ e $\alpha, \beta \in \mathbb R$ temos que

$$L_X(\alpha f + \beta g) = ((\alpha f + \beta g)(x_1),\dots, (\alpha f + \beta g)(x_m)) = (\alpha f(x_1) + \beta g(x_1),\dots,\alpha f(x_m) + \beta g(x_m))$$

E como para quaisquer funções em $\mathcal F$ vale que $(f+g)(x) = f(x) + g(x)$, então a expressão é equivalente a

$$(\alpha f(x_1),\dots,\alpha f(x_m)) + (\beta g(x_1),\dots,\beta g(x_m)) = \alpha (f(x_1),\dots,f(x_m)) + \beta(g(x_1),\dots,g(x_m))\\ = \alpha L_X(f) + \beta L_X(g)$$

### Item c

Considere a matriz de Vandermonde

$$V = \begin{bmatrix}
1 & x_{1} & \cdots & x^d_{1}\\
1 & x_{2} & \dots & x^d_{2} \\
\vdots & \vdots && \vdots \\
1 & x_{m} & \dots & x^d_{m}
\end{bmatrix}$$

Assim, dado um polinômio $p = a_0 + a_1 x + \dots + a_d x^d \in\mathcal P_d$, teremos que $L_X(p) = V a$, onde $a = (a_0,\dots,a_d)$. Logo, para calcular $F_{X,d}(b)$ precisamos minimizar $||Va - b||^2_2$ variando $a$. Como V é de posto completo (pela questão 1-e) podemos encontrar a solução $a$ para esse problema de mínimos quadrados calculando $a = V^+b = (V^TV)^{-1}V^Tb$ (Obs: dada uma matriz A, considere que $A^T$ é sua transposta conjugada), logo, conseguimos calcular $R_{X,d}$ usando a base canônica de $\mathcal P_d$ fazendo $R_{X,d} = (V^TV)^{-1}V^T$.

### Item d

Observe que, como $\kappa(L_X) = \kappa(V)$, bem como $\kappa(R_{X,d}) = \kappa(V^+) = \kappa(V)$*, isso intuitivamente nos mostra que à medida que o $d$ aumenta, os valores $1,x,\dots,x^d$ das colunas de $V$ podem ficar numéricamente muito voláteis quando aplicado um $x_i$ qualquer. Isso pois, caso $|x_i|<1$ teremos $x^d\rightarrow 0$ e (principalmente) caso $|x_i|>1$ teremos que $x^d$ cresce muito rapidamente (deixando as colunas da matriz $V$ desbalanceadas, com $x^k$ cada vez mais distantes de $x^{k-1}$), assim $d\uparrow$ $\Rightarrow \kappa(V)\uparrow$.

Já com relação ao aumento de $m$, ainda de modo intuitivo, conforme aumenta-se a quantidade de pontos (deixando $d$ fixo) é esperado que a interpolação de torne mais estável, já que com mais dados do que parâmetros (e considerando, em específico, que $m >> d$) o sistema se torna superdeterminado, consequentemente, estabilizando o número de condicionamento.

*Veja que, se $V^+$ é a pseudoinversa de $V$, então dado que $V=U\Sigma W^T$, devemos ter que $V^+=W\Sigma^+ U^T$ para que $V^+V=I$, onde $\Sigma=$ diag $(\sigma_1,\dots,\sigma_{d+1})$ e $\Sigma=$ diag $(\frac{1}{\sigma_1},\dots,\frac{1}{\sigma_{d+1}})$. Portanto

$$\kappa(V^+) = \frac{\sigma_{max}(V^+)}{\sigma_{min}(V^+)}=\frac{\frac{1}{\sigma_{min}(V)}}{\frac{1}{\sigma_{max}(V)}}=\frac{\sigma_{max}}{\sigma_{min}}=\kappa(V)$$

### Item e:

Sendo $(p_1,\dots,p_{d+1})$ a base dos polinômios interpoladores de Lagrange, como cada $p_i(x_j) = \delta_{ij}$ para $j\in${$1,\dots,d+1$} e valores $a_{ij}$ quaisquer para os outros pontos $x_{d+2},\dots,x_m$, temos que $L_X(p_i)=(p_i(x_1),\dots,p_i(x_i),\dots,p_i(x_{d+1}),p_i(x_{d+2}),\dots,p_i(x_m))=(0,\dots,1,\dots,0,a_{(d+2)i},\dots,a_{mi})$, portanto, nessa base, $L_X = \begin{bmatrix}I_{d+1}\\A\end{bmatrix}_{m\times (d+1)}$ onde $A$ é uma matriz $(m-d-1)\times(d+1)$ cujos elementos são $A_{ij}=p_j(x_{d+1+i})$. Assim, por construção, teremos que

$$R_{X,d} = (\begin{bmatrix}I_{d+1}&A^T\end{bmatrix}\begin{bmatrix}I_{d+1}\\A\end{bmatrix})^{-1}\begin{bmatrix}I_{d+1}&A^T\end{bmatrix} = \begin{pmatrix}I_{d+1}+A^TA\end{pmatrix}^{-1}\begin{bmatrix}I_{d+1}&A^T\end{bmatrix}=\begin{bmatrix}(I+A^TA)^{-1}&(I+A^TA)^{-1}A^T\end{bmatrix}$$

Uma matriz por blocos, donde segue valendo que $R_{X,d}$ é uma inversa à esquerda de $L_X$*.

Com relação ao condicionamento, como o bloco $A$ tem cada uma das suas entradas sendo não nula (se alguma fosse, os polinômios $p_i$ teriam ao menos $d+1$ valores $x_j$ diferentes que são raízes, um absurdo, pois os polinômios tem grau $d$) ao calcularmos $L^T_XL_X = \begin{bmatrix}I_{d+1}&A^T\end{bmatrix}\begin{bmatrix}I_{d+1}\\A\end{bmatrix} = I_{d+1} + B^TB$ a matriz $B^TB$ será simétrica e positiva semidefinida e, portanto, terá autovalores $\lambda_i\geq 0$. Assim, os autovalores de $L^T_XL_X$ serão $1 +\lambda_i$ e, logo, $L_X$ terá valores singulares $\sqrt{1+\lambda_i} \geq 1$. Enfim, o número de condicionamento será $\kappa(L_X) = \frac{\sqrt{1+\lambda_{max}}}{\sqrt{1+\lambda_{min}}} \geq 1$ (e como mostrado antes, o mesmo condicionamento vale para $R_{X,d}$), com seu crescimento dependendo exclusivamente do quão distantes entre si os autovalores de $B$ ficam. 

*$R_{X,d}L_X = \begin{bmatrix}(I+A^TA)^{-1}&(I+A^TA)^{-1}A^T\end{bmatrix}\begin{bmatrix}I_{d+1}\\A\end{bmatrix} = (I+A^TA)^{-1}(I+A^TA) = I_{d+1}$.

(Obs: ademais, para cálculo da base dos polinômios de lagrange, dado um conjunto $X$ de pontos, calculou-se cada coluna $j$ por $L_j(x)=\prod^d_{k\neq j}\frac{x-x_k}{x_j-x_k}$, onde qualquer polinômio $p\in\mathcal P_d$ pode ser representado nesta base por uma combinação linear $\sum^d_{j=0}\alpha_jL_j$)

```python
def base_lagrange(x_i, x, j):
    d = len(x) - 1
    y = np.ones_like(x_i, dtype=float)
    for k in range(d + 1):
        if k != j:
            y *= (x_i - x[k]) / (x[j] - x[k])
    return y

def lagrange_L(X, d):
    m = len(X)
    x = np.linspace(X.min(), X.max(), d + 1)
    L = np.zeros((m, d + 1))
    for j in range(d+1):
        L[:, j] = base_lagrange(X, x, j)
    return L

def lagrange_R_comp_L(x, deg):
    lx = lagrange_L(x, deg)
    r = np.linalg.inv(lx.T @ lx) @ lx.T
    return r, lx

X = np.linspace(-1, 1, 20) # 10 pontos de 0 a 10
d = 5 # Polinômio de grau 5

R, L = lagrange_R_comp_L(X, d)

# Calculando os números de condicionamento (norma 2)
cond_L = np.linalg.cond(L)
cond_R = np.linalg.cond(R)

print(f"Condicionamento de L_X:       {cond_L:.2e}")
print(f"Condicionamento de R_X,d:   {cond_R:.2e}")
```

### Item f

Para a base canônica, temos que a matriz de $L_X$ é construída tendo que calcular cada um dos elementos $(L_X)_{ij} = x^j_i$, isto é, fazer $m*(d+1)$ cálculos, logo, a complexidade é $O(md)$. Já com matriz de $R_{X,d}$ é necessário realizar os cálculos do produto $V^TV$ (que é $O(md^2)$, já que é uma multiplicação de matrizes), de inverter $V^TV$ (que é $O(d^3)$, se feita por eliminação gaussiana) e de multiplicar por $V^T$ (mais $O(md^2)$), portanto o custo total é $O(md^2 + d^3)$. E como visto anteriormente, para esta base, o condicionamento de $L_X$ e $R_{X,d}$ podem crescer arbitrariamente a depender dos valores $m$ e $d$.

Por outro lado, para a base dos polinômios interpoladores de Lagrange, a construção da matriz $L_X$​ exige a avaliação de cada entrada $(L_X​)_{ij}​=L_j​(x_i​)$. Como cada polinômio da base de Lagrange é definido por um produtório de $d$ termos, o cálculo de cada elemento da matriz leva $O(d)$ operações. Como a matriz possui tamanho $m×(d+1)$, o custo total de construção de $L_X$​ salta para $O(md^2)$. O cálculo da matriz RX,d​, segue envolvendo a computação de $(L^T_X​L_X​)^{−1}L^T_X​$, o que mantém a complexidade em $O(md^2+d^3)$

```python
import time

m = 500
graus_d = np.arange(2, 45)
X = np.linspace(-1, 1, m)
tempos_canon = []
tempos_lagr = []
conds_canon = []
conds_lagr = []
runs = 10 # número de repetições para tirar a média dos tempos

for d in graus_d:
    # base canonica
    t0 = time.perf_counter()
    for _ in range(runs):
        L_c = L_X(X, d)
    t1 = time.perf_counter()
    tempos_canon.append((t1 - t0) / runs)
    conds_canon.append(np.linalg.cond(L_c))
    # base de lagrange
    t0 = time.perf_counter()
    for _ in range(runs):
        L_l = lagrange_L(X, d)
    t1 = time.perf_counter()
    
    tempos_lagr.append((t1 - t0) / runs)
    conds_lagr.append(np.linalg.cond(L_l))

tempos_canon = np.array(tempos_canon)
tempos_lagr = np.array(tempos_lagr)
conds_canon = np.array(conds_canon)
conds_lagr = np.array(conds_lagr)
# calculando a eficiência de cada base com (velocidade) / (condicionamento)
razao_canon = 1.0 / (tempos_canon * conds_canon)
razao_lagr = 1.0 / (tempos_lagr * conds_lagr)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
# tempos de execução
ax1.plot(graus_d, tempos_canon * 1000, label="Canônica")
ax1.plot(graus_d, tempos_lagr * 1000, label="Lagrange")
ax1.set_xlabel("Valor de d")
ax1.set_ylabel("Tempo computacional")
ax1.legend()
# eficiência
ax2.plot(graus_d, razao_canon, label="Canonica")
ax2.plot(graus_d, razao_lagr, label="Lagrange")
ax2.set_xlabel("Valor de d")
ax2.set_ylabel("Eficiência")
ax2.set_yscale('log')
ax2.legend()
plt.tight_layout()
plt.show()
```

Assim vemos que de fato a complexidade computacional do cálculo de $L_X$ na base de Lagrange é maior do que na base canônica. No entanto, ao analisar a eficiência do algoritmo (velocidade de cálculo pela estabilidade numérica) na base de Lagrange se mostra superior a base canônica, isto é, o baixo condicionamento compensa o custo extra de processamento.

### Item a:

Estamos supondo $n \geq d + 1$. Veja que a aplicação composta $L_Y\circ R_{X,d}$ possui dimensões $n \times m$, uma vez que calculamos $L_Y = V$ (que tem dimensões $n \times (d+1)$) e $F_{X,d} = (V^TV)^{-1}V^T$ (que tem dimensões $(d+1) \times m$). Perceba que $Im(R_{X,d}) = \mathcal P_d$, pois dado qualquer polinômio $p\in\mathcal P_d$, tomando $b = L_X(p)$, temos $R_{X,d}(b) = p$, assim, a composição $L_Y\circ R_{X,d}$ possui imagem $Im(L_Y)$ (considerando $L_Y: \mathcal P_d \rightarrow \mathbb R^n$) e, como $L_Y$ tem posto cheio nas colunas, $dim(Im(L_Y)) = d + 1$. Portanto, $rank(L_Y\circ R_{X,d}) = d+1$.

### Item b:

Sejam $B_1​$ e $B_2$ bases quaisquer, existe uma matriz de mudança de base $M$ de dimensão $(d+1)×(d+1)$, inversível, que converte os vetores de $B_1$​ para $B_2​$. Sendo assim, tome $B_1$ como a base canônica e $B_2$ a base de Lagrange, teremos que podemos mudar a base da matriz de avaliação por $L_{X,B_2}=L_{X,B_1}M^{-1}$ e $L_{Y,B_2}=L_{Y,B_1}M^{-1}$. Defina ainda $C_2=L_{Y,B_2}R_{Y,d,B_2}$ a aplicação da composta na base lagrangiana. Das propriedades das operações de matrizes e lembrando que $R=(L^TL)^{-1}L^T$, temos que

$$R_{Y,d,B_2}=((L_{X,B_1}M^{-1})^T(L_{X,B_1}M^{-1}))^{-1}(L_{X,B_1}M^{-1})^T=(M^{-T}L_{X,B_1}L_{X,B_1}M^{-1})^{-1}M^{-T}L_{X,B_1}=\\(M(L_{X,B_1}L_{X,B_1})^{-1}M^T)M^{-T}L_{X,B_1}=M(L^T_{X,B_1}L_{X,B_1})=MR_{X,d,B_1}$$

Agora podemos substituir na equação

$$C_2=L_{Y,B_2}R_{Y,d,B_2}=(L_{Y,B_1}M^{-1})(MR_{X,d,B_1})=L_{Y,B_1}R_{X,d,B_1}=C_1$$

Portanto, independentemente da base a transformação composta $L_Y\circ R_{X,d}$ é identica.

(Obs: de um modo muito ingênuo e talvez falacioso também poderia-se pensar que como $L_Y\circ R_{X,d}: \mathbb R^m \rightarrow \mathbb R^n$, não dependeria da base utilizada no meio do processo)

### Item c:

Pois ambos os conjuntos de pontos $X$ e $Y$ estão contidos no mesmo intervalo $[-1,1]$. Quando aplicamos $R_{X,d}$, encontramos um polinômio que se ajusta ao conjunto de pontos dados em $X$ e, então, quando aplicamos $L_Y$ nesse polinômio geramos um novo conjunto de pontos em $Y$. Contudo, como ambos os conjuntos estão contidos no mesmo intervalo, bem como são gerados pela mesma correlação envolvendo primeiras e segundas coordenas, isto é, a relação de serem funções iguais diferindo por pequenas perturbações nos valores (barulho).

De uma outra forma, é como se estivéssemos criando um método que, com base em dados de teste anteriores gerasse uma previsão dos próximos pontos. Nesse sentido, podemos definir o processo como uma interpolação.

### Item d

Ao analisarmos os valores singulares da matriz de aplicação composta em ambas as bases, observamos uma queda drástica na magnitude dos valores singulares após o ($d+1$)-ésimo valor singular (teoricamente, como a dimensão de $\mathcal P_d$ é $d+1$, deveriam ser estritamente 0, inclusive), o que nos dá um $\sigma_{min}$ extremamente pequeno (computacionalmente próximo de $10^{-16}$, por conta de erros de aproximação numérica), gerando um condicionamento que explode para o infinito. 

Para além disso, comparando as matrizes geradas na base canônica e na base de Lagrange, observamos também que os primeiros $d+1$ valores singulares são idênticos, corroborando com as conclusões chegadas no item b, de que essa igualdade entre as bases se sustenta quando $d$ é comportadinho.

```python
def calculate_phi(x, degree):

    phi = np.zeros((len(x), degree + 1))

    for i in range(degree + 1):
        phi[:, i] = x ** i

    return phi

def lr_pi_pol(x, y, degree):

    phi = calculate_phi(x, degree)

    w = np.linalg.solve(phi.T @ phi, phi.T @ y)

    return w, phi

def base_lagrange(x_i, x, j):
    d = len(x) - 1
    y = np.ones_like(x_i, dtype=float)
    for k in range(d + 1):
        if k != j:
            y *= (x_i - x[k]) / (x[j] - x[k])
    return y

def lagrange_L(X, d):
    m = len(X)
    x = np.linspace(X.min(), X.max(), d + 1)
    L = np.zeros((m, d + 1))
    for j in range(d+1):
        L[:, j] = base_lagrange(X, x, j)
    return L

def lagrange_R_comp_L(x, deg):
    lx = lagrange_L(x, deg)
    r = np.linalg.inv(lx.T @ lx) @ lx.T
    return r, lx



#variando o n somente
m=10
n=10
d=10
cond_n=[]
cond_n_lag=[]
for n in range(10,400):
    x=np.linspace(-1,1,m)
    y=np.linspace(-1,1,n)
    lx=calculate_phi(x,d)
    rx=np.linalg.pinv(lx)
    ly=calculate_phi(y,d)
    ly_rx=ly@rx

    lx=lagrange_L(x,d)
    ly_rx_lag=lagrange_L(y,d)@np.linalg.pinv(lx)
    singular_values=np.linalg.svd(ly_rx,compute_uv=False)
    singular_values_lag=np.linalg.svd(ly_rx_lag,compute_uv=False)

    cond_n.append(singular_values.max()/singular_values.min())
    cond_n_lag.append(singular_values_lag.max()/singular_values_lag.min())



ms=np.arange(10,400)
plt.plot(ms,cond_n,label="base caoninca")
plt.plot(ms,cond_n_lag, label="lagrange")
plt.legend()
plt.grid()

plt.show()





#variando o m somente
m=10
n=200
d=10
cond_m=[]
cond_m_lag=[]
for m in range(10,400):
    x=np.linspace(-1,1,m)
    y=np.linspace(-1,1,n)
    lx=calculate_phi(x,d)
    rx=np.linalg.pinv(lx)
    ly=calculate_phi(y,d)
    ly_rx=ly@rx
    singular_values=np.linalg.svd(ly_rx,compute_uv=False)
    lx=lagrange_L(x,d)
    ly_rx_lag=lagrange_L(y,d)@np.linalg.pinv(lx)
    singular_values_lag=np.linalg.svd(ly_rx_lag,compute_uv=False)

    cond_m.append(singular_values.max()/singular_values.min())
    cond_m_lag.append(singular_values_lag.max()/singular_values.min())
ms=np.arange(10,400)

plt.yscale("log")
plt.plot(ms,cond_m,label="base canonica")
plt.plot(ms,cond_m_lag,label="lagrange")
plt.grid()
plt.legend()
plt.show()






#variando o d somente
m=200
n=200
d=10
cond_d=[]
cond_d_lag=[]
for d in range(10,400):
    x=np.linspace(-1,1,m)
    y=np.linspace(-1,1,n)
    lx=calculate_phi(x,d)
    rx=np.linalg.pinv(lx)
    ly=calculate_phi(y,d)
    ly_rx=ly@rx
    
    lx=lagrange_L(x,d)
    ly_rx_lag=lagrange_L(y,d)@np.linalg.pinv(lx)
    singular_values=np.linalg.svd(ly_rx,compute_uv=False)
    singular_values_lag=np.linalg.svd(ly_rx_lag,compute_uv=False)
    cond_d.append(singular_values.max()/singular_values.min())
    cond_d_lag.append(singular_values_lag.max()/singular_values_lag.min())
ms=np.arange(10,400)

plt.yscale("log")
plt.plot(ms,cond_d)
plt.plot(ms,cond_d_lag)
plt.grid()

plt.show()
```

### Item g

Ao considerar a composta $L_Y\circ R_X$ para um conjunto $Y\in[-1.5,1.5]$ somos capazes de aproximar a função aos pontos do intervalo interior $X\in[-1,1]$ com precisão (a depender do valor de $d+1$), porém aplicada à região exterior ($Y\backslash X$) estamos realizando uma extrapolação nos intervalos $[-1.5,-1)$ e $(1,1.5]$, explodindo o erro relativo à proximidade da função aos pontos. E esse comportamento é consequência dos termos de maior grau do polinômio, que crescem (ou decrescem) violentamente fora do intervalo que originalmente foram ajustados para aproximar os pontos.

```python
def mean_squared_error(y_true,y_pred):
    return (((y_true-y_pred)**2).sum())/len(y_true)

rng = np.random.default_rng(0)

m=150
d=10
n=200
x_train=np.linspace(-1,1,m)
x_test=np.linspace(-1.5,1.5,n)
y_train=np.sin(8*x_train)+0.05*rng.normal(0,1,m)
y_test=np.sin(8*x_test)+0.05*rng.normal(0,1,n)

r, lx=R_comp_L(x_train,y_train,d)
index_dentro=np.where((x_test>=-1)&(x_test<=1))[0]
x_dentro=x_test[index_dentro]
y_dentro=y_test[index_dentro]
index_fora=np.where((x_test<-1)|(x_test>1))[0]
x_fora=x_test[index_fora]
y_fora=y_test[index_fora]

pol=np.polynomial.Polynomial(w)
y_pred_train=pol(x_train)
y_pred_dentro=pol(x_dentro)
print("erro entre valor predizido e valor real dentro de [-1,1]: ",mean_squared_error(y_dentro,y_pred_dentro))
y_pred_fora=pol(x_fora)
print("erro entre valor predizido e valor real fora de [-1,1]: ",mean_squared_error(y_fora,y_pred_fora))

plt.plot(x_train,y_train, label="Função Verdadeira: sen(8x)")
plt.plot(x_train,y_pred_train, label=f"Extrapolação Polinomial (d={d})")
plt.xlabel("Intervalo")
plt.ylabel("Erro?")
plt.grid()
plt.show()
y_pred_test=pol(x_test)
plt.plot(x_test,y_test)
plt.plot(x_test,y_pred_test)
plt.xlabel("Intervalo")
plt.ylabel("Erro?")
plt.grid()
plt.show()
```
