### Item a:
Dado $p\in\mathcal P_d$ um polinômio qualquer, defina $b_p = (p(x_1),p(x_2),\dots,p(x_m))\in\mathbb R^m$ como o vetor obtido por $L_{X}(p)$. Além disso, seja $q = R_{X,d}(b)\in\mathcal P_d$ o polinômio que minimiza $||L_X(q) - b_p||^2_2$, mas como $b_p = L_X(p)$, tomando $q = p$, segue que

$$||L_X(q) - b_p||^2_2 = ||L_X(q) - L_X(p)||^2_2 = ||L_X(p) - L_X(p)||^2_2 = 0$$

Portanto, $p$ é um minimizador, basta mostrar que é único. Para isso, provemos que $N(L_X) = \set 0$ em $\mathcal P_d$.

Suponha por absurdo que exista $p\neq 0 \in\mathcal P_d$ tal que $L_X(p) = 0$, assim $(p(x_1),\dots,p(x_m)) = (0,\dots,0)$, ou seja, $p(x_i) = 0, \forall i \in$ {$1,\dots, m$}, o quê implica que $p$ possui $m$ raízes, mas isso é um absurdo, pois $p$ é um polinômio de grau até $d < d+1\leq m$. Logo, como o núcleo de $L_X : \mathcal P_d \rightarrow \mathbb R^m$ é trivial, $L_X$ é injetiva e, portanto, o minimizador é o único.

Com isso, temos que

$$||L_X(q) - L_X(p)||^2_2 = 0 \Leftrightarrow L_X(q) - L_X(p) = 0 \Leftrightarrow L_X(q) = L_X(p) \Leftrightarrow q = p$$

Assim, por fim, temos que $R_{X,d}(L_X(p)) = p$ e como isso vale para todo $p \in \mathcal P_d$, segue que $R_{X,d}\circ L_X = \mathcal Id_{\mathcal P_d}.$

### Item b:

Basta provar que $R_{X,d}$ é linear na soma e na multiplicação por escalar.

De fato, temos que $R_{X,d}(b + v) = R_{X,d}(b) + R_{X,d}(v)$, pois dados $b,v \in \mathbb R^m$ defina $p = R_{X,d}(b)$ e $q = R_{X,d}(v)$ polinômios de $\mathcal P_d$, defina ainda $h = p + q$. Assim, uma vez que $L_X$ é linear*, $L_X(h) = L_X(p) + L_X(q)$, logo $R_{X,d}(L_X(h)) = R_{X,d}(L_X(p) + L_X(q))$, mas como, pelo item a, $R_{X,d}(L_X(h)) = h = p + q = R_{X,d}(L_X(p)) + R_{X,d}(L_X(q))$, segue que é linear para soma.

Ademais, dado $\alpha \in \mathbb R$, temos que

$$R_{X,d}(\alpha b) = R_{X,d}(L_X(\alpha p) = \alpha p = \alpha R_{X,d}(L_X(p)) = \alpha R_{X,d}(b))$$

Logo também é linear para a multiplicação por escalar.

*$L_X$ é linear pois dados $f,g \in \mathcal F$ e $\alpha, \beta \in \mathbb R$ temos que

$$L_X(\alpha f + \beta g) = ((\alpha f + \beta g)(x_1),\dots, (\alpha f + \beta g)(x_m)) = (\alpha f(x_1) + \beta g(x_1),\dots,\alpha f(x_m) + \beta g(x_m))$$

E como para quaisquer funções em $\mathcal F$ vale que $(f+g)(x) = f(x) + g(x)$, então a expressão é equivalente a

$$(\alpha f(x_1),\dots,\alpha f(x_m)) + (\beta g(x_1),\dots,\beta g(x_m)) = \alpha (f(x_1),\dots,f(x_m)) + \beta(g(x_1),\dots,g(x_m)) = \alpha L_X(f) + \beta L_X(g)$$
